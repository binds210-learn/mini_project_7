```python
df_clean = df.copy()
```

```python
df_clean.columns = (df_clean.columns
                    .str.strip()
                    .str.lower()
                    .str.replace(" ", "_")
                    .str.replace("/", "_")
                    .str.replace("'", "")
                    .str.replace("(", "", regex=False)
                    .str.replace(")", "", regex=False))
```

```python
df_clean = df_clean.rename(columns={"nacionality": "nationality"})
```
